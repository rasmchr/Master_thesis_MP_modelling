```python
import numpy as np
import matplotlib.pyplot as plt

z= np.linspace(0, 150, 1000) 
D1 = 1.0*10**(-3) / np.sqrt(2*np.pi) * np.exp(-(z - 50)**2)
plt.figure(figsize=(10,6))
plt.plot(D1, z, label='Diffusion Coefficient')
plt.gca().invert_yaxis()
plt.xlabel("Diffusion Coefficient")
plt.axhline(y=50, linestyle='--',label="Seabed")
plt.legend()
plt.ylabel("Depth (m)")
plt.grid()
plt.show()
```
