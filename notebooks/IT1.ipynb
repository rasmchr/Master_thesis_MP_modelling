```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
u = 10.0  # velocity
t = np.linspace(0, 5, 100)  # time range

# Different starting points (z0 values)
z0_values = np.linspace(-20, 50, 15)

# Plot characteristics
plt.figure()

for z0 in z0_values:
    z = u * t + z0
    plt.plot(t, z, label=f"z0={z0:.1f}")

# Plot the moving front z = ut
plt.plot( t,u*t, 'k--', linewidth=2, label="z = ut (front)")

# Labels and styling
plt.xlabel("t")
plt.ylabel("z")
#plt.title("Characteristic Curves: z = ut + z0")
plt.grid(True)
# 👉 Move legend outside
plt.legend(
    loc='center left',
    bbox_to_anchor=(1, 0.5),
    fontsize=8
)

# 👉 Add space so it doesn't get cut off
plt.tight_layout()

plt.show()
```
