```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
alpha = 1.74
lambda_ = np.sqrt(alpha)

# Velocity (same as your model, downward = negative)
def u(z):
    return 10**(-2) + 3.66*10**4 / (1 + np.exp(z - 50))

# Source Damköhler number
def Da_source(z):
    return alpha * np.exp(-lambda_ * z) / np.abs(u(z))  
    # use abs(u) to represent transport magnitude

# Depth range
z_vals = np.linspace(0, 150, 500)

# Compute
Da_vals = Da_source(z_vals)

# Plot
plt.figure(figsize=(6,5))
plt.semilogx(Da_vals, z_vals)

plt.gca().invert_yaxis()
plt.xlabel(r"$\mathrm{DA}_{\mathrm{source}}(z)$")
plt.ylabel("Depth z")
plt.title("Source Damköhler number vs depth")
plt.grid()

plt.show()
```
